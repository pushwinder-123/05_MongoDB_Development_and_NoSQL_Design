```python
import pandas as pd
from pymongo import MongoClient
```

```python
client = MongoClient(
    "mongodb+srv://pushwinderd0001:pushwinderd0001@cluster0.uwhxwvf.mongodb.net/"
)

db = client["northstar_operational_analytics_db"]
```

```python
customers = pd.read_csv(
    "https://raw.githubusercontent.com/pushwinder-123/northStar_Dataset/refs/heads/main/customers.csv"
)

deliveries = pd.read_csv(
    "https://raw.githubusercontent.com/pushwinder-123/northStar_Dataset/refs/heads/main/deliveries.csv"
)

complaints = pd.read_csv(
    "https://raw.githubusercontent.com/pushwinder-123/northStar_Dataset/refs/heads/main/complaints.csv"
)

app_events = pd.read_csv(
    "https://raw.githubusercontent.com/pushwinder-123/northStar_Dataset/refs/heads/main/app_events.csv"
)
```

```python
customers_data = customers.to_dict("records")

deliveries_data = deliveries.to_dict("records")

complaints_data = complaints.to_dict("records")

app_events_data = app_events.to_dict("records")
```

```python
customers_collection = db["customers"]

deliveries_collection = db["deliveries"]

complaints_collection = db["complaints"]

app_events_collection = db["app_events"]
```

```python
customers_collection.insert_many(customers_data)

deliveries_collection.insert_many(deliveries_data)

complaints_collection.insert_many(complaints_data)

app_events_collection.insert_many(app_events_data)
```

CRUD Operations in MongoDB
CREATE Operation

```python
new_customer = {
    "customer_id": "C9999",
    "age": 30,
    "home_zone": "CENTRAL",
    "customer_type": "Premium",
    "loyalty_score": 90,
    "account_status": "Active"
}

customers_collection.insert_one(new_customer)
```

READ Operation

```python
active_customers = customers_collection.find(
    {"account_status": "Active"}
)

for customer in active_customers:
    print(customer)
```

UPDATE Operation

```python
customers_collection.update_one(
    {"customer_id": "C9999"},
    {
        "$set": {
            "loyalty_score": 95
        }
    }
)
```

DELETE Operation

```python
customers_collection.delete_one(
    {"customer_id": "C9999"}
)
```

MongoDB Query 1 — High Severity Complaints

```python
high_complaints = complaints_collection.find(
    {"severity": "High"}
)

for complaint in high_complaints:
    print(complaint)
```

MongoDB Query 2 — Failed Deliveries

```python
failed_deliveries = deliveries_collection.find(
    {"delivery_status": "Failed"}
)

for delivery in failed_deliveries:
    print(delivery)
```

MongoDB Query 3 — Application Events with High API Latency

```python
high_latency = app_events_collection.find(
    {"api_latency_ms": {"$gt": 500}}
)

for event in high_latency:
    print(event)
```

MongoDB Query 4 — Customers with High Loyalty Scores

```python
loyal_customers = customers_collection.find(
    {"loyalty_score": {"$gt": 80}}
)

for customer in loyal_customers:
    print(customer)
```

MongoDB Query 5 — Aggregation Query

```python
pipeline = [
    {
        "$group": {
            "_id": "$severity",
            "total_complaints": {"$sum": 1}
        }
    }
]

result = complaints_collection.aggregate(pipeline)

for data in result:
    print(data)
```

Aggregation Query  — High-Risk Operational Analysis

```python
pipeline = [

    {
        "$group": {

            "_id": "$delivery_status",

            "average_override_count": {
                "$avg": "$manual_route_override_count"
            },

            "average_customer_rating": {
                "$avg": "$customer_rating_post_delivery"
            },

            "total_deliveries": {
                "$sum": 1
            }
        }
    },

    {
        "$sort": {
            "average_override_count": -1
        }
    }
]

results = deliveries_collection.aggregate(pipeline)

for data in results:
    print(data)
```
